## Génération automatique de profils utilisateurs

Ce notebook génère un ensemble de profils utilisateurs fictifs à partir des données d’images disponibles dans `annotations.json`.

Chaque utilisateur est défini par :
- Un identifiant (`user_1`, `user_2`, …)
- Une ou plusieurs couleurs préférées (issues des couleurs dominantes d’images)
- Un format d’image préféré (`JPEG`, `PNG`, etc.)
- Une orientation préférée (`portrait`, `landscape`, `square`)
- Une liste d’images aimées (tirées aléatoirement du dataset)

L’objectif est de simuler un ensemble de profils afin de tester le système de recommandation.

Les profils sont enregistrés automatiquement dans `user_profiles.json`.


```python
import json
import random
import pandas as pd

ANNOTATION_FILE = "../annotation/annotations.json"
USER_PROFILES_FILE = "JSON/user_profiles.json"

# Fonction utilitaire pour charger un fichier JSON
def load_json(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        return json.load(f)

# Charger les annotations des images
annotations = load_json(ANNOTATION_FILE)

# Générer un ensemble de profils utilisateurs fictifs
def generate_users(num_users=5):
    users = []
    image_keys = list(annotations.keys())  # Liste des noms d'image disponibles

    for i in range(1, num_users + 1):
        user = {
            "id": f"user_{i}",
            "preferred_colors": random.choice([img.get("dominant_colors", []) for img in annotations.values()]),
            "preferred_format": random.choice([img.get("metadata", {}).get("format", "JPEG") for img in annotations.values()]),
            "preferred_orientation": random.choice(["portrait", "landscape", "square"]),
            "liked_images": random.sample(image_keys, min(5, len(image_keys)))
        }
        users.append(user)
    
    return users

# Sauvegarder les profils dans un fichier JSON
def save_user_profiles(num_users=5):
    users = generate_users(num_users=num_users)
    with open(USER_PROFILES_FILE, "w", encoding="utf-8") as f:
        json.dump(users, f, indent=4, ensure_ascii=False)
    return users

# Génération effective
df_users = pd.DataFrame(save_user_profiles(num_users=5))
display(df_users.head())
```

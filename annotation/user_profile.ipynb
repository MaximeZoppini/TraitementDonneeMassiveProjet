```python
import json
import random
import pandas as pd

ANNOTATION_FILE = "../data_collection/annotations.json"
USER_PROFILES_FILE = "../annotation/user_profiles.json"

def load_json(file_path):
    with open(file_path, "r") as f:
        return json.load(f)

annotations = load_json(ANNOTATION_FILE)

def generate_users(num_users=5):
    users = []
    image_keys = list(annotations.keys())  # Liste des images disponibles

    for i in range(1, num_users + 1):
        user = {
            "id": f"user_{i}",
            "preferred_colors": random.choice([img["dominant_colors"] for img in annotations.values()]),
            "preferred_format": random.choice(["JPEG", "PNG"]),
            "preferred_orientation": random.choice(["portrait", "landscape"]),
            "liked_images": random.sample(image_keys, min(3, len(image_keys)))  # Sélection aléatoire d'images aimées
        }
        users.append(user)
    
    return users

def save_user_profiles():
    users = generate_users(num_users=5)  # Modifier le nombre si besoin
    with open(USER_PROFILES_FILE, "w") as f:
        json.dump(users, f, indent=4)

    return users

user_profiles = save_user_profiles()

df_users = pd.DataFrame(user_profiles)
display(df_users.head())
```

# Chargement et Structuration des Données  

Ce notebook permet de **charger les métadonnées des images** et de les organiser sous forme de **DataFrame Pandas**.  
Nous allons :  
- les informations des images depuis `annotations.json`  
- Transformer les données en **tableau structuré**  
- Vérifier que tout est bien chargé  


## 📦 Installation des dépendances  
Avant de commencer, nous vérifions que la bibliothèque `pandas` est bien installée.

```
pip install pandas
```


````python
import json
import pandas as pd

ANNOTATION_FILE = "annotations.json"

def load_annotations():
    with open(ANNOTATION_FILE, "r") as f:
        data = json.load(f)
    
    records = []
    for image, details in data.items():
        record = {
            "image": image,
            "format": details["metadata"]["format"],
            "size_x": details["metadata"]["size"][0],
            "size_y": details["metadata"]["size"][1],
            "mode": details["metadata"]["mode"],
            "colors": ", ".join(details["dominant_colors"])
        }
        records.append(record)
    
    return pd.DataFrame(records)

if __name__ == "__main__":
    df = load_annotations()
    print(df.head())
````
